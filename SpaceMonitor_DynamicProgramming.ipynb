## Lógica da Solução

Inicialmente são criadas 35 áreas monitoradas.

Cada área é conectada a outras áreas através de um grafo.

O algoritmo calcula o menor custo para percorrer as áreas monitoradas.

Para evitar cálculos repetidos, é utilizada uma técnica chamada memoização.

A memoização armazena resultados já calculados em memória. Quando o mesmo cálculo é solicitado novamente, o sistema utiliza o valor armazenado em vez de recalcular.

Essa técnica é um dos conceitos fundamentais da Programação Dinâmica.


```python
# ==========================================
# SpaceMonitor - Dynamic Programming
# ==========================================

# memória para programação dinâmica
memo = {}

# criação do grafo
grafo = {}

# cria 35 áreas monitoradas
for i in range(1, 36):
    grafo[f"Area{i}"] = {}

# cria conexões entre áreas
for i in range(1, 35):
    grafo[f"Area{i}"][f"Area{i+1}"] = i + 1


# função para calcular menor caminho
def menor_caminho(area):

    if area in memo:
        return memo[area]

    if len(grafo[area]) == 0:
        return 0

    menor = float("inf")

    for vizinho, custo in grafo[area].items():

        resultado = custo + menor_caminho(vizinho)

        if resultado < menor:
            menor = resultado

    memo[area] = menor

    return menor


# função para exibir rotas
def mostrar_rotas():

    print("===== SPACEMONITOR =====\n")

    for area in grafo:

        print(
            f"{area} -> custo mínimo para atendimento: {menor_caminho(area)}"
        )


# execução
mostrar_rotas()
```

Foi utilizado um grafo para representar as áreas monitoradas por satélites.

Cada área monitorada corresponde a um vértice do grafo.

As conexões entre áreas representam possíveis rotas de comunicação ou deslocamento.
